### 사칙연산 계산기

```python
num1 = float(input("첫번째 숫자를 입력하세요: "))
num2 = float(input("두번째 숫자를 입력하세요: "))
operator = input("연산자를 선택하세요(+,-,*,/): ")

def calculate(num1, num2, operator):
    if operator == '+':
        return num1 + num2
    elif operator == '-':
        return num1 - num2
    elif operator == '*':
        return num1 * num2
    elif operator == '/':
        if num2 == 0:
            return "Division by zero"
        return num1 / num2
    else:
        return "Invalid operator"
    
    
result = calculate(num1, num2, operator)
print("결과: ", result)
```
